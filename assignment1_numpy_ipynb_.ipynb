Q49. Save a NumPy array into a .npy file using np.save(). Load the saved file using np.load() and
display its contents.

```python
import numpy as np

# Create an array
a = np.array([10, 20, 30, 40, 50])

# Save the array
np.save("my_array.npy", a)

# Load the saved array
b = np.load("my_array.npy")

# Display the contents
print(b)
```
