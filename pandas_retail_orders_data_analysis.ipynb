**Problem_Statement4:Find the product category with the maximum frequency of Cancelled orders and create a DataFrame showing category-wise cancelled order frequency.**

```python
cancelled=df[df['OrderStatus']=='Cancelled']
```

```python
cancelled
```

```python
result = cancelled.groupby("Category").size().reset_index(name="Frequency")
```

```python
result
```

```python
result[result["Frequency"]==result["Frequency"].max()]
```
